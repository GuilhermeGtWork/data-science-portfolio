# Análise Exploratória - Dataset Tips (Gorjetas) - Importação

Neste projeto, vamos analisar o comportamento de gorjetas em um restaurante,
explorando padrões por dia, horário e tamanho da mesa.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## Importação da Base de Dados

O dataset a seguir que será utilizado, contém informações sobre gorjetas em um restaurante, e suas informações Gerais.

```python
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv")
df.head()

df.info()
```

### Observação inicial
O Dataset possui 244 registros com 7 variáveis sem valores faltando, duas colunas para valores decimais(`float`), uma coluna para valores inteiros(`int`) e 4 colunas para valores do tipo Texto (`string`).

```python
df.describe()
```

### Informações iniciais das Gorjetas
A tabela mostra que o valor médio da conta (`total_bill`) é de aproximadamente 19.79, variando entre 3.07 e 50.81  
A gorjeta média (`tip`) é de cerca de 2.99, com valores entre 1.00 e 10.00.  
Já o tamanho médio dos grupos (`size`) é de aproximadamente 2.56 pessoas por mesa, variando de 1 até 6 clientes.

# Gráfico - Clientes por Dia 
   Nesta Etapa, vamos analisar a quantidade de clientes por dia nesse restaurante,
   explorando o padrão e frequência nos quatro dias mais relevantes da semana.

```python

df["day"].value_counts().plot(kind="bar")
plt.title("Quantidade de clientes por dia")
plt.xlabel("Dia da semana")
plt.ylabel("Número de clientes")
plt.show()
```

### Insight

Podemos ver que o dia com mais Clientes da semana foi Sábado(`Saturday`),
indicando maior movimento nos fins de semana.

```python
dia_mais_movimentado = df["day"].value_counts().idxmax()
print("O dia com mais clientes é:", dia_mais_movimentado)
print("com a quantidade de clientes:", df["day"].value_counts().max())
```

## Gorjeta média por dia


```python
df.groupby("day")["tip"].mean().plot(kind="bar")
plt.title("Gorjeta média por dia")
plt.xlabel("Dia da semana")
plt.ylabel("Gorjeta média")
plt.show()
```

### Insight
A gorjeta média é maior no domingo e no sábado, sugerindo que nos fins de semana os clientes tendem a contribuir mais em gorjetas, possivelmente por maior consumo ou eventos de lazer.

## Gorjeta por tamanho da mesa


```python
df.groupby("size")["tip"].mean().plot(kind="bar")
plt.title("Gorjeta média por tamanho do grupo")
plt.xlabel("Tamanho do grupo")
plt.ylabel("Gorjeta média")
plt.show()
```

### Insight
Podemos observar que, quanto maior o tamanho da mesa, maior tende a ser o valor médio da gorjeta,
demonstrando um padrão crescente para grupos maiores, o que sugere um maior retorno em gorjetas nesses casos.

## Conclusão

Neste projeto, foi possível identificar padrões de movimento no restaurante,
com maior concentração de clientes aos finais de semana e suas gorjetas.
Essas informações podem ajudar no planejamento operacional.
